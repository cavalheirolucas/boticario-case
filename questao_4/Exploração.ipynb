```python
#Exploração das bases de dados

import pandas as pd

df = pd.read_csv('base_vendas_boticario.csv')
meta = pd.read_csv('tabela_metas_mensais_estados.csv')
```

```python
#Checagem de tamanho e tipos de dados (Base de dados Metas)
print(meta.shape)
print(meta.dtypes)
```

```python
#Checagem dos dados nas colunas (Base de dados Metas)
print(meta['Estado'].unique())
print('------------')
print(meta['Ano'].unique())
print('------------')
print(meta['Mês'].unique())
```

```python
#Checando dados de Mês e ano (Base de dados Metas)
print(meta.groupby('Ano')['Mês'].nunique())
```

```python
#Validando dados nulos (Base de dados Metas)
print(meta.isnull().sum())
```

```python
#Checagem de tamanho e tipos de dados (Base de dados Vendas)
print(df.shape)
print(df.dtypes)
```

```python
#Validando dados nulos (Base de dados Vendas)
print(df.isnull().sum())
```

```python
# Checando Intervalo de data (Base de dados Vendas)
print(df['Data'].min())
print(df['Data'].max())
```

```python
#Checando concentração de vendas por canal e estado (Base de dados Vendas)
print(df['Canal'].value_counts())
print('----------------')
print(df['Estado'].value_counts())
```

```python
#Checando dados de cidade (Base de dados Vendas)
print(sorted(df['Cidade'].unique()))
```

```python
#Checando os dados de quantidade (Base de dados Vendas)
print(df['Quantidade'].describe())
print('------------')
print(df['Quantidade'].sort_values(ascending=False).head(10))
```

```python
#Checando concentração de valores nas quantidades (Base de dados Vendas)
soma_por_valor = df.groupby('Quantidade')['Quantidade'].count().sort_values(ascending=False)
print(soma_por_valor.head(10))
print('------------')
#Checando concentração de valores nos canais e por quantidades (Base de dados Vendas)
soma_por_valor_canal = df.groupby(['Canal','Quantidade'])['Quantidade'].count().sort_values(ascending=True)
print(soma_por_valor_canal)
```

```python
#Checando os dados de Valor Unitário (Base de dados Vendas)
print(df['Valor_Unitario'].describe())
print('------------')
print(df['Valor_Unitario'].sort_values(ascending=False).head(10))
```

```python
#Checando os dados de Custo Unitário (Base de dados Vendas)
print(df['Custo_Unitario'].describe())
print('------------')
print(df['Custo_Unitario'].sort_values(ascending=False).head(10))
```

```python
#Checando valores das colunas (Base de dados Vendas)
for col in ['Categoria_Produto', 'Tipo_pagamento', 'Status_Entrega', 'Cupom_Utilizado','Genero_Cliente']:
    print(df[col].unique())
    print('------------')
```

```python
#Checando intervalo de datas de nascimento (Base de dados Vendas)
print(df['Data_Nascimento_Cliente'].min())
print(df['Data_Nascimento_Cliente'].max())
```

```python
#Checando qtd dados de feedback (Base de dados Vendas)
print(df['Feedback_Cliente'].value_counts(dropna=False))
```

```python
#Checando qtd dados de cupom (Base de dados Vendas)
print(df['Cupom_Utilizado'].value_counts(dropna=False))
```

```python
#Checando dados duplicados
print(df.duplicated().sum())
```

```python
#Checando incosistencia nos dados de valores
neg_margem = df[df['Custo_Unitario'] > df['Valor_Unitario']]
print(len(neg_margem))
```

```python
#Checando cruzamento de dados entre as tabelas
print('Em vendas mas não em metas:', set(df['Estado'].unique()) - set(meta['Estado'].unique()))
print('Em metas mas não em vendas:', set(meta['Estado'].unique()) - set(df['Estado'].unique()))
```
